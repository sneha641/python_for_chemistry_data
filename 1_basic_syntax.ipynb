```python
#Lists

# This is a list
energy_kcal = [-13.4, -2.7, 5.4, 42.1]
print(energy_kcal)
energy_length = len(energy_kcal)
print(energy_length)
print( 'The length of list is', energy_length)

#print first element of list
print(energy_kcal[0])
#Convert the second list element to kilojoules.
energy_kjoules = energy_kcal[1] * 4.184
print(energy_kjoules)
```

```python
#slice

short_list = energy_kcal[0:2]
print(short_list)

slice1 = energy_kcal[1:]
slice2 = energy_kcal[:3]
print('slice1 is', slice1)
print('slice2 is', slice2)
#my answere 
# slice 1 is [-2.7, 5.4, 42.1]
# slice 2 is [-13.4, -2.7, 5.4]

#loop

for number in energy_kcal:
    kj = number * 4.184
    print(kj)
    
#append

energy_kJ = []
for number in energy_kcal:
    kJ = number * 4.184
    energy_kJ.append(kJ)
    
print(energy_kJ)
    
```

```python
# logic statements

negative_energy_kJ = []
for number in energy_kJ:
    if number < 0:
        negative_energy_kJ.append(number)
print(negative_energy_kJ)

negative_numbers = []
for number in energy_kJ:
    if number < 0 or number == 0:
        negative_numbers.append(number)
print(negative_numbers)
```

```python
#test
#Set up a for loop to go over each element of data_list. 
# If the element is a string (str), recast it as a float. 
# Save all of the numbers to a new list called number_list

data_list = ['-12.5', 14.4, 8.1, '42']
number_list = []
for element in data_list:
    if type(element) is str:
        element = float(element)
    number_list.append(element)

print(number_list)
```
