```python
import os
import numpy

distance_file = os.path.join ('data', 'distance_data_headers.csv')
distance = numpy.genfromtxt(fname=distance_file, delimiter=',', dtype='unicode')
headers= distance[0]
data = distance[1:]
data = data.astype(float)
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(data[:,1])
```

```python
sample = headers[1]
plt.figure()
plt.xlabel('simulation frame')
plt.ylabel('Distance(angstromg)')
fig_1 = plt.plot(data[:,1], label=sample)
plt.legend()
plt.savefig ( f'{sample}.png' )
```

```python
plt.figure()
plt.xlabel('simulation frame')
plt.ylabel('Distance(angstromg)')
plt.plot(data[:,1], label=headers[1])
plt.plot(data[:,2], label=headers[2])
plt.legend()

#plt.savefig('two_sample.png')
```

```python
for col in range(1, len(headers)):
    fig = plt.plot(data[:,col], label=headers[col])
    plt.legend()
```

Different plot for each sample and save

```python
for col in range(1, len(headers)):
    plt.figure()
    sample = headers[col]
    plt.plot(data[:,col], label=sample)
    plt.legend()
    plt.savefig( F'{sample}.png')
    
```

calculate max and min limits for plot

```python
data_min = numpy.min(data[:, 1:])
data_max = numpy.max(data[:, 1:])
print(data_min)
print(data_max)
```

To make the plots all have the same y axis range consider scaling minimum and maximum to give for data

```python
data_min = numpy.min(data[:,1:])*0.9
data_max = numpy.max(data[:,1:])*1.1

for col in range(1, len(headers)):
    plt.figure()
    sample = headers[col]

    plt.plot(data[:,col], label=sample)
    plt.ylim(data_min, data_max)
    plt.legend()
    plt.xlabel('Simulation Frame')
    plt.ylabel('Distance (angstrom)')

    plt.savefig(F'{sample}.png')
```

To get every other row to the 10th row, start of 0, and end of 10, and an increment of 2. We then use the : to get every column.

```python
print(data[0:10:2,:])
```

How would you get every 100th point in the first column?

```python
incr_100 = (data[0::100,0])
print(incr_100)
```
