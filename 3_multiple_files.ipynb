03_Prod.mdout this is a file output by the Amber molecular dynamics simulation program. Assignment is to parse this file, and write a new file containing a list of the > total energies. Name your file Etot.txt

```python
# Build file path
filename = os.path.join('data', '03_prod.mdout')

# Open file for reading
f = open(filename,'r')

# Read the data in the file
data = f.readlines()

#close file
f.close()

# open a file for writing
f_write = open('Etot.txt', 'wt')

# Loop through the lines in the file
for line in data:
    split_line = line.split()
    #print(split_line)
    if 'Etot' in line:
        print(split_line[2])
        # Get information from the line
        f_write.write(f'{split_line[2]}\n')
f_write.close()
```
