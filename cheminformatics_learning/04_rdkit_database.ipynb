###Using RDKit to Build a Dataset

There is a file called amino_acids.txt in the data folder that contains SMILES for the 20 naturally occurring amino acids. Build a dataset for molecules and write a file. Create a comma-separated value file with columns SMILES, num_heavy (number of heavy atoms), molecular_weight, and one other molecular descriptor for the molecules in the file.

```python
#Read SMILES from txt file

with open('data/amino_acids.txt', 'r') as f:
    smiles_list = [line.strip() for line in f if line.strip()] 
    
    #for line in f — loops through each line in the file.
    #line.strip() — removes whitespace/newline characters (\n) from both ends of the line.

print(f"Read {len(smiles_list)} SMILES")
for s in smiles_list:
    print(s)
```

```python
#Make an RDKit molecule for each SMILES
from rdkit import Chem

molecules = [Chem.MolFromSmiles(s) for s in smiles_list]

print(f"Created {len(molecules)} molecules")
for i, mol in enumerate(molecules):
    if mol is None:
        print(f"  SMILES {i}: INVALID - {smiles_list[i]}")
    else:
        print(f"  SMILES {i}: {smiles_list[i]}")
```

```python
#Get the number of heavy atoms, molecular weight, and LogP for each molecule
from rdkit.Chem import Descriptors

num_heavy = [mol.GetNumHeavyAtoms() for mol in molecules]
molecular_weight = [Descriptors.MolWt(mol) for mol in molecules]
logP = [Descriptors.MolLogP(mol) for mol in molecules]

for i in range(len(smiles_list)):
    print(f"{smiles_list[i]:45s}  Heavy: {num_heavy[i]:3d}  MW: {molecular_weight[i]:7.2f}  LogP: {logP[i]:6.2f}")
```

```python
#Create amino_acids_processed.csv with data
import csv

with open('data/amino_acids_processed.csv', 'w', newline='') as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(['SMILES', 'num_heavy', 'molecular_weight', 'LogP'])
    for i in range(len(smiles_list)):
        writer.writerow([smiles_list[i], num_heavy[i], f"{molecular_weight[i]:.2f}", f"{logP[i]:.2f}"])

print("Saved amino_acids_processed.csv")
```
