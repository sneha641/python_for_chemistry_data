Question
Use the provided sapt.out file. In this output file, the program calculates the interaction energy for an ethene-ethyne complex. The output reports four interaction energy components: electrostatics, induction, exchange, and dispersion. Parse each of these energies, in kcal/mole, from the output file. (Hint: study the file in a text editor to help you decide what to search for.) Calculate the total interaction energy by adding the four components together. Your code’s output should look something like this:

Electrostatics : -2.25850118 kcal/mol
Exchange : 2.27730198 kcal/mol
Induction : -0.5216933 kcal/mol
Dispersion : -0.9446677 kcal/mol
Total Energy : 1.4475602000000003 kcal/mol

````python
import os

sapt_file = os.path.join('data', 'sapt.out')
print(sapt_file)
````

````python
outfile = open(sapt_file, "r")
saptout = outfile.readlines()
print(len(saptout))

````

````python
for line in saptout:
    if 'Electrostatics' in line and 'sSAPT0' not in line:
        electro_line = line
        print(electro_line)
        
    if 'Exchange' in line and 'sSAPT0' not in line:
        exchange_line = line
        print(exchange_line)
    if 'Induction' in line and 'sSAPT0' not in line:
        induction_line = line
        print(induction_line)
    if 'Dispersion' in line and 'sSAPT0' not in line:
        dis_line = line
        print(dis_line)
````

````python
electro_energy = float(electro_line.split()[3])
exchange_energy = float(exchange_line.split()[3])
induction_energy = float(induction_line.split()[3])
dispersion_energy = float(dis_line.split()[3])

total_energy = electro_energy + exchange_energy + induction_energy + dispersion_energy

print(f'Electrostatics : {electro_energy} kcal/mol')
print(f'Exchange : {exchange_energy} kcal/mol')
print(f'Induction : {induction_energy} kcal/mol')
print(f'Dispersion : {dispersion_energy} kcal/mol')
print(f'Total Energy : {total_energy} kcal/mol')
````
