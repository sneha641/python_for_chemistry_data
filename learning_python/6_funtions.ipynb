 Write a function called open_xyz which takes an xyz file as a parameter and returns the symbols and coordinates.

```python
def open_xyz(file_name):
    xyz_file = numpy.genfromtxt(fname=file_name, skip_header=2, dtype='unicode')
   
    symbols = xyz_file[:,0]
    coord = xyz_file[:,1:]
    coord = coord.astype(float)
    return symbols, coord
```

```python
import os
import numpy

def distance_coordinates(atom1coord, atom2coord):
    
    #calculate distance between 2D points
    
    x_distance = atom1coord[0] - atom2coord[0]
    y_distance = atom1coord[1] - atom2coord[1]
    z_distance = atom1coord[2] - atom2coord[2]
    bond_length = numpy.sqrt(x_distance**2 + y_distance**2 + z_distance**2) 
    return bond_length
```

```python
def bond_check(atom_distance, minimum_distance=0, maximum_distance=1.5):
    
    if atom_distance > minimum_distance and atom_distance <= maximum_distance:
        return True
    else:
        return False
```

```python
def print_bonds(atom_symbols, atom_coordinates):
    """
    Prints atom symbols and bond length for a set of atoms.
    """
    num_atoms = len(atom_symbols)
    for num1 in range(0, num_atoms):
        for num2 in range(0, num_atoms):
            if num1 < num2:
                bond_length_12 = calculate_distance(atom_coordinates[num1], atom_coordinates[num2])
                if bond_check(bond_length_12) is True:
                    print(F'{atom_symbols[num1]} to {atom_symbols[num2]} : {bond_length_12:.3f}')
```

# Matrix Multiplication (3x1 and 1x3)

A **3×1** matrix (column vector) multiplied by a **1×3** matrix (row vector) gives a **3×3** matrix (outer product).

A **1×3** matrix (row vector) multiplied by a **3×1** matrix (column vector) gives a **1×1** scalar (dot product).

```python
import numpy as np

# 3x1 column vector
A = np.array([[1],
              [2],
              [3]])
print("A (3x1):")
print(A)
print("Shape:", A.shape)

# 1x3 row vector
B = np.array([[4, 5, 6]])
print("\nB (1x3):")
print(B)
print("Shape:", B.shape)

# 3x1 @ 1x3 = 3x3 (outer product)
result_3x3 = A @ B
print("\nA @ B (3x1 × 1x3 = 3x3):")
print(result_3x3)
print("Shape:", result_3x3.shape)

# 1x3 @ 3x1 = 1x1 (dot product / scalar)
result_1x1 = B @ A
print("\nB @ A (1x3 × 3x1 = 1x1):")
print(result_1x1)
print("Shape:", result_1x1.shape)
```
